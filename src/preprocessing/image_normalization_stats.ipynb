```python
import os
import cv2
import numpy as np
from tqdm import tqdm
```

```python
# Set your image folder path (e.g., normal images)
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
DATA_DIR = os.path.join(ROOT_DIR, 'data')
image_dir = os.path.join(DATA_DIR, 'train', 'normal')
image_files = [f for f in os.listdir(image_dir) if f.endswith('.jpg')]
```

```python
# Accumulators
means = []
stds = []
```

```python
for file in tqdm(image_files, desc='Processing images'):
    img = cv2.imread(os.path.join(image_dir, file))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = img / 255.0 # normalize to [0, 1]

    means.append(np.mean(img, axis=(0,1)))
    stds.append(np.std(img, axis=(0,1)))
```

```python
# Convert to numpy
means = np.array(means)
stds = np.array(stds)
```

```python
# Average across all images
mean_rgb = np.mean(means, axis=0)
std_rgb = np.mean(stds, axis=0)
```

```python
print("Mean (R, G, B):", mean_rgb)
print("Std  (R, G, B):", std_rgb)
```
