```python
import cv2
import os
import numpy as np
```

```python
scale = 4
```

```python
def scale_image(img, img_name, scale):
    h, w = img.shape[:2]
    
    new_w = max(1, w // scale)
    new_h = max(1, h // scale)

    low_res = cv2.resize(img, (new_w, new_h))    
    return low_res
```

```python
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
in_path = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced')
out_path = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced normal')

files = os.listdir(in_path)
```

```python
for image_name in files:
    img = cv2.imread(os.path.join(in_path, image_name))
    low_res_img = scale_image(img, image_name, scale)

    if low_res_img is None:
        continue

    filename = f'{out_path}/{image_name}'
    cv2.imwrite(filename, low_res_img)
```
