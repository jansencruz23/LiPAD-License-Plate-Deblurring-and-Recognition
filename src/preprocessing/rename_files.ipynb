```python
import os
import shutil
```

```python
src_dir = os.path.abspath('C:\\Thesis\\LiPAD with Paddle\\src\\gan\\low_light\\results')
dest_dir = os.path.abspath('C:\\Thesis\\LiPAD with Paddle\\src\\gan\\low_light\\results_renamed')
```

```python
files = os.listdir(src_dir)

for file in files:
    name = file.split('.')
    name = f'low_light_{name[0]}.jpg'
    shutil.copy2(os.path.join(src_dir, file), os.path.join(dest_dir, name))
```
