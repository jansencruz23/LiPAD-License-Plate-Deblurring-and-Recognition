```python
import os
import shutil
import random
from tqdm import tqdm
```

```python
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
DATA_DIR = os.path.join(ROOT_DIR, 'data')
class_dirs = ['enhanced', 'h_blur', 'low_light', 'low_qual', 'normal', 'v_blur']
train_dir = os.path.join(DATA_DIR, 'train')
valid_dir = os.path.join(DATA_DIR, 'valid')
```

```python
# Get all image filenames
all_files = os.listdir(os.path.join(DATA_DIR, 'normal'))
all_files = [f for f in all_files if f.endswith('.jpg')]
all_files.sort()
random.shuffle(all_files)
```

```python
# Split
split_idx = int(len(all_files) * 0.9)
train_files = all_files[:split_idx]
valid_files = all_files[split_idx:]
```

```python
# Helper function to copy files
def copy_files(filenames, target_dir):
    for class_name in class_dirs:
        os.makedirs(os.path.join(target_dir, class_name), exist_ok=True)
        for fname in tqdm(filenames, desc=f'Copying to {target_dir}/{class_name}'):
            src_path = os.path.join(DATA_DIR, class_name, fname)
            dst_path = os.path.join(target_dir, class_name, fname)
            if os.path.exists(src_path):
                shutil.copy2(src_path, dst_path)
```

```python
copy_files(train_files, train_dir)
copy_files(valid_files, valid_dir)
```
