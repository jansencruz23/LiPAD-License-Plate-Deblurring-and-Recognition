```python
import os
import cv2
import shutil
```

```python
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
in_path = os.path.join(ROOT_DIR, 'data', 'raw - 17k')
out_path = os.path.join(ROOT_DIR, 'data', 'output 17k')
```

```python
files = os.listdir(in_path)
unique_files = {}
scale_factor = 1.65
```

```python
for file in files:
    base_name = os.path.splitext(file)[0]  # Remove file extension if any
    parts = base_name.split('_')
    
    if len(parts) >= 2:
        key = f"{parts[0]}_{parts[1]}"
    else:
        key = base_name  # Fallback if there's only one part

    if key not in unique_files:
        unique_files[key] = file
```

```python
for file in files:
    if "BURST" in file:
        # Get the common prefix before BURST number
        prefix = file.split("BURST")[0] + "BURST"
    else:
        prefix = file  # fallback

    # Store only the first file seen for each prefix
    if prefix not in unique_files:
        unique_files[prefix] = file
```

```python
for file in unique_files.values():
    src = os.path.join(in_path, file)
    dst = os.path.join(out_path, file)
    shutil.move(src, dst)
```

### Resize

```python
for image_file in files:    
    img_path = os.path.join(in_path, image_file)
    img = cv2.imread(img_path)

    height, width = img.shape[:2]

    if height <= 640 or width <= 640:
        new_width = int(width * scale_factor)
        new_height = int(height * scale_factor)
        img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_LINEAR)
        print(f"Resized: {image_file} to {new_width}x{new_height}")

    out_file = os.path.join(out_path, image_file)
    cv2.imwrite(out_file, img)
```

```python
import os
import cv2
import shutil
```

```python
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
in_dir  = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'normal')
out_dir = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'normal filtered 2')

os.makedirs(out_dir, exist_ok=True)
```

```python
min_size = 128
max_size = 400

for fname in os.listdir(in_dir):
    fpath = os.path.join(in_dir, fname)
    img = cv2.imread(fpath)

    if img is None:
        continue

    h, w = img.shape[:2]

    if h > max_size or w > max_size:
        continue

    out_path = os.path.join(out_dir, fname)
    shutil.copy(fpath, out_path)
```

```python
ROOT_DIR = os.path.abspath(os.path.join(os.getcwd(), '..', '..'))
in_dir  = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced 2x')
out_dir = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced filtered 2')

os.makedirs(out_dir, exist_ok=True)
```

```python
folder1 = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced 2x')
folder2 = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'normal filtered 2')
folder3 = os.path.join(ROOT_DIR, 'data', 'all', 'experiment2', 'train', 'enhanced filtered 2')

files1 = set(os.listdir(folder1))
files2 = set(os.listdir(folder2))

common_files = files1.intersection(files2)

for filename in common_files:
    src_path = os.path.join(folder1, filename)
    dst_path = os.path.join(folder3, filename)
    shutil.copyfile(src_path, dst_path)
```
