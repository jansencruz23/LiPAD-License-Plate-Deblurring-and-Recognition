```python
import os
import shutil
```

```python
in_path = os.path.join('data','train')
out_path = os.path.join('data','train out')
```

```python
files = os.listdir(in_path)
unique_files = {}
```

```python
for file in files:
    base_name = file.split('.jpg')[0]
    key = base_name.split('_jpg')[0]
    if key not in unique_files:
        unique_files[key] = file
```

```python
for file in unique_files.values():
    src = os.path.join(in_path, file)
    dst = os.path.join(out_path, file)
    shutil.move(src, dst)
```
